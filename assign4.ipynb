```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('/content/Most Streamed Artists on Spotify (17_07_2026) V1.1.csv')
display(df.head())
```

```python

top_artists = df.sort_values(by='Total Streams (in millions)', ascending=False).head(10)
sns.barplot(data=top_artists, x='Total Streams (in millions)', y='Artist Name', palette='viridis')
plt.title('Top 10 Most Streamed Artists on Spotify')
plt.xlabel('Total Streams (in millions)')
plt.ylabel('Artist Name')
plt.tight_layout()
plt.show()
```

```python

sns.scatterplot(
    data=df,
    x='Debut Year',
    y='Total Streams (in millions)',
    hue='Primary Genre',
    alpha=0.8,
    s=100
)
plt.title('Total Streams vs. Debut Year by Primary Genre')
plt.xlabel('Debut Year')
plt.ylabel('Total Streams (in millions)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python

artist_col = ' Artist Type' if ' Artist Type' in df.columns else 'Artist Type'
artist_counts = df[artist_col].value_counts()

plt.pie(artist_counts, labels=artist_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribution of Artist Types')
plt.tight_layout()
plt.show()
```

```python

sns.histplot(df['Total Streams (in millions)'], bins=15, kde=True, color='purple')
plt.title('Histogram of Total Streams')
plt.xlabel('Total Streams (in millions)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python

yearly_streams = df.groupby('Debut Year')['Total Streams (in millions)'].mean().reset_index()
sns.lineplot(data=yearly_streams, x='Debut Year', y='Total Streams (in millions)', marker='o', color='b')
plt.title('Average Total Streams by Debut Year')
plt.xlabel('Debut Year')
plt.ylabel('Average Total Streams (in millions)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python

top_genres = df['Primary Genre'].value_counts().head(5).index
filtered_df = df[df['Primary Genre'].isin(top_genres)]
sns.boxplot(data=filtered_df, x='Primary Genre', y='Total Streams (in millions)', palette='Set2')
plt.title('Total Streams Distribution Across Top Primary Genres')
plt.xlabel('Primary Genre')
plt.ylabel('Total Streams (in millions)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
