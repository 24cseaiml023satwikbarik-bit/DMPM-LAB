```python
import pandas as pd
import numpy as np


df_anscombe = pd.read_json('/content/sample_data/anscombe.json')
df = df_anscombe[df_anscombe['Series'] == 'I'][['X', 'Y']]

print("Data loaded from anscombe.json (Dataset I). First 5 rows:")
display(df.head())
```

```python
correlation_matrix = df.corr()
covariance_matrix = df.cov()

correlation_xy = correlation_matrix.loc['X', 'Y']
covariance_xy = covariance_matrix.loc['X', 'Y']

print(f"Correlation Coefficient between X and Y: {correlation_xy:.4f}")
print(f"Covariance between X and Y: {covariance_xy:.4f}")
```
