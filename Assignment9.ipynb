# Question1

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori

dataset = [
    ["bread", "milk"],
    ["bread", "diaper", "beer", "eggs"],
    ["milk", "diaper", "beer", "coke"],
    ["bread", "milk", "diaper", "beer"],
    ["bread", "milk", "diaper", "coke"],
]


all_items = sorted(list(set(item for row in dataset for item in row)))
encoded_data = [
    {item: (item in transaction) for item in all_items} for transaction in dataset
]
df = pd.DataFrame(encoded_data)

frequent_itemsets = apriori(df, min_support=0.60, use_colnames=True)
frequent_itemsets["length"] = frequent_itemsets["itemsets"].apply(len)

print("Frequent Itemsets:")
print(frequent_itemsets)
```

# Question2

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency


observed = np.array([[23, 18, 8, 8], [15, 6, 12, 10]])
rows = ["Male", "Female"]
cols = ["Strawberry", "Coffee", "Orange", "Vanilla"]


chi2, p, dof, expected = chi2_contingency(observed)


expected_df = pd.DataFrame(np.round(expected, 2), index=rows, columns=cols)


print("a) Hypotheses:")
print("   H0: Flavor preference is independent of gender.")
print("   H1: Flavor preference is dependent on gender.\n")
print(f"b) Degrees of Freedom: {dof}\n")
print("c) Expected Frequency Table:")
print(expected_df)
```
