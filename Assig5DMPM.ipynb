```python
import numpy as np
import pandas as pd
```

```python
arr1=np.array([10,20,30,40,50])
```

Z-score normalization

```python
def z_score(arr):
  mean=np.mean(arr)
  sd=np.std(arr)
  z_score_nor=(arr-mean)/sd
  return z_score_nor
```

min max normalization

```python
def min_max(arr):
  small=arr.min()
  large=arr.max()
  arr_nor=(arr-small)/(large-small)
  return arr_nor
```

```python
def deci(arr):
  j=0
  v=np.max(abs(arr))
  while v>=1:
    v=v/(10**j)
    j+=1
  arr_nor=arr/(10**j)
  return arr_nor
```

```python
df=pd.read_csv("gold_stock.csv")
df['Close'] = pd.to_numeric(df['Close'], errors='coerce')
df.dropna(subset=['Close'], inplace=True)
df
```

```python
print(deci(df['Close']))
```

```python
print(min_max(df['Close']))
```

```python
print(z_score(df['Close']))
print(np.mean(z_score(df['Close'])))
print(np.std(z_score(df['Close'])))
print(np.var(df['Close']))
```
